```
import gradio as gr
```

```
def saludo(nombre, intensidad):
  return "Hola, " + nombre + "!" * int(intensidad)
```

```
demo = gr.Interface(
  fn=saludo,
  inputs=["text", "slider"],
  outputs=["text"],
)

demo.launch(share=True)
```
